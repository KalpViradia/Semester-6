# Decision Tree Classifier

```python
from sklearn.datasets import load_iris
```

# Import iris data set using sklearn

```python
data = load_iris()
```

# Importing the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Divide the data into input and output

```python
X = data['data']
y = data['target']
```

```python
y
```

# Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Fitting DecisionTreeClassifier   on dataset

```python
from sklearn.tree import DecisionTreeClassifier

irisModel = DecisionTreeClassifier()
print(irisModel.fit(X_train, y_train))
```

# Display Decision Tree

```python
from sklearn import tree

%matplotlib qt

tree.plot_tree(irisModel)
```

# Predict the x_test 

```python
irisModel.predict(X_test)
```

```python
irisModel.score(X_test, y_test)
```

# Save Model State

```python
import pickle
```

```python
with open('iris_model.pkl', 'wb') as file:
    pickle.dump(irisModel, file)
```

# Load Model from Disk

```python
with open('iris_model.pkl', 'rb') as file:
    loaded_iris_model = pickle.load(file)
```

```python
loaded_iris_model.predict(X_test)
```

# Import diabetes.csv dataset 

```python
df = pd.read_csv('diabetes.csv')
```

```python
df.head()
```

# Check the distribution of the target

```python
df['Outcome'].value_counts()
```

# Replace zeros with NaN for specific columns

```python
cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
df[cols] = df[cols].replace(0, np.nan)
```

# Check for missing values

```python
df.isnull().sum()
```

# Fill missing values with median

```python
df[cols] = df[cols].fillna(df[cols].median())
```

# Visualize Distributions

```python
sns.pairplot(df, hue='Outcome')
```

# Convert Target data into integer code

```python
df['Outcome'] = df['Outcome'].map({'Non Diabetic': 0, 'Diabetic': 1})
```

```python
df.head()
```

# Divide the data into input and output

```python
X = df.iloc[::, :-1]
y = df['Outcome']
```

# Splitting the dataset into the Training set and Test set

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Create Model

```python
diabetesModel = DecisionTreeClassifier()
```

# Fitting DecisionTreeClassifier   on dataset

```python
print(diabetesModel.fit(X_train, y_train))
```

# Display Decision Tree

```python
tree.plot_tree(diabetesModel)
```

# Predict the x_test 

```python
diabetesModel.predict(X_test)
```

# Display Training Accuracy

```python
diabetesModel.score(X_train, y_train)
```

# Display Test Accuracy

```python
diabetesModel.score(X_test, y_test)
```

# Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

cm  = confusion_matrix(y_test, diabetesModel.predict(X_test))

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Healthy', 'Diabetic'], 
            yticklabels=['Healthy', 'Diabetic'])
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.title('Confusion Matrix Heatmap')
plt.show()
```

# Save Model State

```python
with open('diabetes_model.pkl', 'wb') as file:
    pickle.dump(diabetesModel, file)
```

# Load Model from Disk

```python
with open('diabetes_model.pkl', 'rb') as file:
    loaded_diabetes_model = pickle.load(file)
```

```python
loaded_diabetes_model.predict(X_test)
```
