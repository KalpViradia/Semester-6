```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load Given dataset -  heart.csv

# Step 7: Split Data into Training and Testing Sets
The dataset is split into training and testing sets for model evaluation.

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.iloc[::, :-1]
y = df['target']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Step 8: Train Decision Tree
We train a Decision Tree Classifier on the training data. You have to also check for KNeighborsClassifier and  GaussianNB

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model_dtc = DecisionTreeClassifier()
```

```python
print(model_dtc.fit(X_train, y_train))
```

```python
y_pred = model_dtc.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, y_pred)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knc = KNeighborsClassifier()
```

```python
print(model_knc.fit(X_train, y_train))
```

```python
y_pred = model_knc.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model_gnb = GaussianNB()
```

```python
print(model_gnb.fit(X_train, y_train))
```

```python
y_pred = model_gnb.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

# Cross Validation

```python
from sklearn.model_selection import KFold
```

```python
k = KFold(n_splits=10)
```

```python
X
```

```python
for train_index, test_index in k.split(X):
    X_train, X_test = X.values[train_index], X.values[test_index]
    y_train, y_test = y[train_index], y[test_index]
    model1 = DecisionTreeClassifier()
    model1.fit(X_train, y_train)
    print(model1.score(X_test, y_test))
```

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(model1,X,y,cv=10)
```
