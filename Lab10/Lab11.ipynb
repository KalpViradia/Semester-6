# Part1: OR GATE

## Step Function

```python
import numpy as np

def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0

print("Setup Complete.")
```

## Initialize Variables

```python
X_or = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_or = np.array([0, 1, 1, 1])
```

```python
weights = np.array([0, 0])
bias = 0
lr = 0.5

print(f"Initial Weights: {weights}")
print(f"Initial Bias: {bias}")
```

## Manual Calculation

```python
input_sample = X_or[1]
target_value = y_or[1]

z = np.dot(input_sample, weights) + bias
print(f"Math: ({input_sample[0]} * {weights[0]}) + ({input_sample[1]} * {weights[1]}) + {bias}")
print(f"Sum (z): {z:.2f}")

prediction = step_function(z)
print(f"Prediction: {prediction}")

error = target_value - prediction
print(f"Error: {error}")
```

```python
epochs = 5

print("---TRAINING (OR GATE)---")

for epoch in range(epochs):
    print(f"\nEPOCH-> {epoch + 1}")

    for i in range(4):
        x = X_or[i]
        y = y_or[i]

        z = np.dot(x, weights) + bias
        pred = step_function(z)

        error = y - pred

        if error != 0:
            weights = weights + (lr * error * x)
            bias = bias + (lr * error * 1)
            print(f"-> MISTAKE on {x}. Expexted {y}, got {pred}.")
            print(f"-> UPDATED weights: {x}. Expexted {y}, got {pred}.")
        else:
            print(f"-> Correct on {x}. Output: {pred}")

print("\n--- FINAL PARAMETERS ---")
print("Weights:", weights)
print("Bias:", bias)
```

# Part1: AND GATE

## Initialize Variables

```python
X_and = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_and = np.array([0, 0, 0, 1])
```

```python
weights = np.array([0, 0])
bias = 0
lr = 0.5

print(f"Initial Weights: {weights}")
print(f"Initial Bias: {bias}")
```

## Manual Calculation

```python
input_sample = X_and[1]
target_value = y_and[1]

z = np.dot(input_sample, weights) + bias
print(f"Math: ({input_sample[0]} * {weights[0]}) + ({input_sample[1]} * {weights[1]}) + {bias}")
print(f"Sum (z): {z:.2f}")

prediction = step_function(z)
print(f"Prediction: {prediction}")

error = target_value - prediction
print(f"Error: {error}")
```

```python
epochs = 5

print("--- TRAINING (AND GATE) ---")

for epoch in range(epochs):
    print(f"\nEPOCH -> {epoch + 1}")

    for i in range(4):
        x = X_and[i]
        y = y_and[i]

        z = np.dot(x, weights) + bias
        pred = step_function(z)

        error = y - pred

        if error != 0:
            weights = weights + (lr * error * x)
            bias = bias + (lr * error)

            print(f"-> MISTAKE on {x}. Expected {y}, got {pred}.")
            print(f"-> UPDATED weights: {weights}, bias: {bias}")
        else:
            print(f"-> Correct on {x}. Output: {pred}")

print("\n--- FINAL PARAMETERS ---")
print("Weights:", weights)
print("Bias:", bias)
```
